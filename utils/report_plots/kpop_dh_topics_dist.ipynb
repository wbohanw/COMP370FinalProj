```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
file = r"C:\Users\Default\Desktop\COMP370FinalProj\data\open_coding\manual_annotation_all_movie_posts.csv"

df = pd.read_csv(file)
```

```python
#only take kpop demon hunters movie
kpop_df = df[df['movie_name'] == 'Kpop Demon Hunters']
topic_counts = kpop_df['topics'].value_counts()
topics = topic_counts.index.tolist()
counts = topic_counts.values.tolist()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

# Professional color
bar_color = '#1F4FBD'

# Plot bars
bars = ax.bar(topics, counts, color=bar_color, alpha=0.9, edgecolor='black', linewidth=0.5)

# Title and labels
ax.set_title(
    'Distribution of Discussion Topics in Reddit Posts for K-pop Demon Hunters',
    fontsize=16, fontweight='bold', pad=20, fontname='Times New Roman'
)
ax.set_xlabel('Topics', fontsize=14, labelpad=15, fontname='Times New Roman')
ax.set_ylabel('Number of Posts', fontsize=14, labelpad=15, fontname='Times New Roman')

# Horizontal grid lines
ax.grid(axis='y', linestyle='--', alpha=0.6)


# Remove top and right spines
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

# Adjust y-axis to fit labels
if counts and max(counts) > 0:
    ax.set_ylim(0, max(counts) * 1.15)

# X-axis labels rotation
plt.xticks(rotation=45, ha='right', fontsize=11)

plt.tight_layout()
plt.savefig(r'C:\Users\Default\Desktop\COMP370FinalProj\data\plot_data\kpop_demon_hunters_topics_distribution.png', dpi=300)

plt.show()

```
