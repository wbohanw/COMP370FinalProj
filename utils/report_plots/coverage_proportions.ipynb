```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
annotated_filename = r"C:\Users\Default\Desktop\COMP370FinalProj\data\all_movie_posts.csv"
```

```python
annotated_data = pd.read_csv(annotated_filename)
```

```python
def autopct_func(pct):
    return f"{pct:.1f}%" if pct > 4 else ""
```

```python
# Pie chart of movie proportions
movie_counts = annotated_data['movie_name'].value_counts()
movie_counts = movie_counts.sort_values(ascending=False)

# Professional font
plt.rcParams['font.size'] = 12
plt.rcParams['font.family'] = 'Times New Roman'

# Muted professional palette
colors = [
    "#262D68", "#323B86", "#404BA7", "#665191", "#a05195",
    "#d45087", "#f95d6a", "#ff7c43", "#ffa600",
    "#c7c7c7", "#969696"
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.pie(
    movie_counts,
    radius=0.75,
    labels=movie_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=colors[:len(movie_counts)],
    wedgeprops={'edgecolor': 'white', 'linewidth': 1},
    textprops={'fontsize': 13}
)

# Main title
ax.set_title(
    "K-pop Demon Hunters Generates the Smallest Share of Reddit Posts Among the Sample",
    fontsize=14,
    pad=20,
    weight='bold'
)

ax.axis('equal')

# Save before showing
plt.savefig(
    r"C:\Users\Default\Desktop\COMP370FinalProj\data\plot_data\movie_post_distribution_pie_chart.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
